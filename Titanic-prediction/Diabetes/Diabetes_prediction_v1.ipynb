```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

train = pd.read_csv('train_diabetes.csv')
test = pd.read_csv('test_diabetes.csv')

train['Insulin']
```

```python
train['Insulin'].fillna(train['Insulin'].mean())
test['Insulin'].fillna(train['Insulin'].mean())
```

```python
features = ['FamilyHistory','Insulin','Age']
X_train = train[features]
y_train = train['Diabetes']
X_test = test[features]

model = LogisticRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)

output = pd.DataFrame({'PatientID':test['PatientID'],'Diabetes': predictions})
output.to_csv('Diabetes_prediction_v1.csv', index=False)
```

```python
# 验证准确率

from sklearn.metrics import accuracy_score
# 读取文件
pred = pd.read_csv('Diabetes_prediction_v1.csv')
true = pd.read_csv('test_answer.csv')

# 合并并计算准确率
merged = pred.merge(true, on='PatientID')
accuracy = accuracy_score(merged['Diabetes_x'], merged['Diabetes_y'])

print(f"模型准确率: {accuracy*100:.2f}%")
```
