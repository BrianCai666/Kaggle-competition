```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
```

```python
# 训练模型
train = pd.read_csv('train_diabetes.csv')
features = ['FamilyHistory', 'Insulin', 'Age']

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(train[features].fillna(train[features].median()), train['Diabetes'])

# 预测测试集
test = pd.read_csv('test_diabetes.csv')
test['Diabetes'] = rf.predict(test[features].fillna(train[features].median()))

# 保存结果
output = pd.DataFrame({'PatientID':test['PatientID'],'Diabetes':test['Diabetes']})
output.to_csv('Diabetes_prediction_v5.csv', index=False)
```

```python
# 验证准确率

from sklearn.metrics import accuracy_score
# 读取文件
pred = pd.read_csv('Diabetes_prediction_v5.csv')
true = pd.read_csv('test_answer.csv')

# 合并并计算准确率
merged = pred.merge(true, on='PatientID')
accuracy = accuracy_score(merged['Diabetes_x'], merged['Diabetes_y'])

print(f"模型准确率: {accuracy*100:.2f}%")
```
