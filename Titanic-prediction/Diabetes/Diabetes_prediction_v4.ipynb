```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
```

```python
# 训练模型
train = pd.read_csv('/Kaggle-competition/Diabetes/train_diabetes.csv')
features = ['Gender', 'HbA1c', 'FamilyHistory']

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(train[features].fillna(train[features].median()), train['Diabetes'])

# 预测测试集
test = pd.read_csv('/Kaggle-competition/Diabetes/test_diabetes.csv')
test['Diabetes'] = rf.predict(test[features].fillna(train[features].median()))

# 保存结果
output = pd.DataFrame({'PatientID':test['PatientID'],'Diabetes':test['Diabetes']})
output.to_csv('/Users/caierchang/Desktop/Diabetes_prediction_v4.csv', index=False)
```
